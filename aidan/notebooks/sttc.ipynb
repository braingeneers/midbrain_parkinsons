```python
def plot_sttc_network(sd, sttc_matrix, line_scale=5.0, max_line_width=10.0,
                      scatter_size=50, title="STTC Network", output_path=None):
    """
    Plot neuron positions with lines connecting each pair, where the line thickness
    is proportional to the STTC value

    Parameters:
    -----------
      sd : SpikeData object
          Contains neuron positions. We'll use get_neuron_positions(sd) to retrieve them.
      sttc_matrix : np.ndarray or dict
          - If it's a 2D NumPy array of shape (N, N), sttc_matrix[i, j] is the STTC for neurons (i, j).
          - If it's a dict with keys = (i, j) pairs, then sttc_matrix[(i, j)] is the STTC value.
      line_scale : float
          A base scaling factor for the line thickness. The actual line width is line_scale * sttc_value.
      max_line_width : float
          Maximum line width (to prevent extremely thick lines).
      scatter_size : float
          Size of the scatter plot markers for neurons.
      title : str
          Title for the figure (also used in the filename if output_path is provided).
      output_path : str or None
          If provided, the plot is saved to that directory with a filename derived from 'title'.
          Otherwise, the plot is shown interactively.
    """
    from analysis.position_sd import get_neuron_positions
    
    # Retrieve positions
    positions = get_neuron_positions(sd)
    n_neurons = positions.shape[0]
    
    # Create figure
    fig, ax = plt.subplots(figsize=(10, 8))
    
    # Plot all neurons as green dots
    ax.scatter(positions[:, 0], positions[:, 1], color='green', s=scatter_size)
    
    # If sttc_matrix is a NumPy array
    if isinstance(sttc_matrix, np.ndarray):
        # We'll assume it's shape (N, N)
        for i in range(n_neurons):
            for j in range(i+1, n_neurons):
                val = sttc_matrix[i, j]
                if val > 0.0:
                    # line width is scaled by val
                    lw = line_scale * val
                    lw = min(lw, max_line_width)
                    start = positions[i]
                    end = positions[j]
                    ax.plot([start[0], end[0]], [start[1], end[1]],
                            color='gray', linewidth=lw, alpha=0.7)
    else:
        # sttc_matrix is a dict with keys=(i, j)
        for (i, j), val in sttc_matrix.items():
            if val > 0.0:
                lw = line_scale * val
                lw = min(lw, max_line_width)
                start = positions[i]
                end = positions[j]
                ax.plot([start[0], end[0]], [start[1], end[1]],
                        color='gray', linewidth=lw, alpha=0.7)
                

    # Create dummy legend handles for specific STTC values.
    dummy1 = mlines.Line2D([], [], color='gray', linewidth=line_scale*0.3,
                             label='STTC = 0.3')
    dummy2 = mlines.Line2D([], [], color='gray', linewidth=line_scale*0.5,
                             label='STTC = 0.5')
    dummy3 = mlines.Line2D([], [], color='gray', linewidth=line_scale*0.8,
                             label='STTC = 0.8')
    
    # Append the dummy handles to the legend.
    handles, labels = ax.get_legend_handles_labels()
    handles.extend([dummy1, dummy2, dummy3])
    labels.extend([dummy1.get_label(), dummy2.get_label(), dummy3.get_label()])
    ax.legend(handles, labels, loc='best')
    
    
    ax.set_title(f"STTC Network:{title}")
    ax.set_xlabel("X Position (µm)")
    ax.set_ylabel("Y Position (µm)")
    

    # If you want to invert y-axis for image-likelihood orientation
    # ax.invert_yaxis()
    
    plt.tight_layout()
    
    # Save or show
    if output_path:
        os.makedirs(output_path, exist_ok=True)
        safe_title = title.replace(" ", "_").replace(":", "")
        filename = f"sttc_network_{safe_title}.png"
        save_path = os.path.join(output_path, filename)
        plt.savefig(save_path)
        plt.close(fig)
        print(f"Saved STTC network plot to {save_path}")
    else:
        plt.show()

```

```python
def keep_top_n_offdiagonal(matrix, n=100):
    """
    Retain only the top n values (by value) in the upper triangle (off-diagonals) of the matrix,
    ignoring any entries equal to 1, and set the rest to 0.
    
    Parameters:
      matrix : np.ndarray
          A square NumPy array (e.g., STTC values).
      n : int, optional
          Number of highest values to retain (default=100).
          
    Returns:
      new_matrix : np.ndarray
          A matrix of the same shape with only the top n off-diagonal (upper triangle) values retained.
    """
    M = matrix.copy()
    N = M.shape[0]
    # Get the indices for the upper triangle (excluding the diagonal)
    triu_idx = np.triu_indices(N, k=1)
    
    # Extract candidate values from the upper triangle.
    candidates = M[triu_idx]
    
    # For selection, ignore values equal to 1 by setting them to -infinity.
    candidates_for_selection = candidates.copy()
    candidates_for_selection[candidates_for_selection == 1] = -np.inf
    
    # If there are fewer than n candidates, just keep them all.
    n_keep = min(n, candidates_for_selection.size)
    
    # Get the indices (in the flattened candidate array) of the top n_keep values.
    top_indices = np.argpartition(candidates_for_selection, -n_keep)[-n_keep:]
    
    # Create a new matrix with zeros.
    new_matrix = np.zeros_like(M)
    
    # Get the corresponding row and column indices from the upper triangle.
    selected_rows = triu_idx[0][top_indices]
    selected_cols = triu_idx[1][top_indices]
    
    # Place the original matrix values into these positions.
    new_matrix[selected_rows, selected_cols] = M[selected_rows, selected_cols]
    
    return new_matrix


def count_entries_above_threshold(matrix, threshold):
    """
    Count the number of elements in a matrix that exceed a given threshold.
    
    Parameters:
      matrix : np.ndarray
      threshold : float
      
    Returns:
      count : int
    """
    return np.count_nonzero(matrix > threshold)
```

```python
def plot_sttc_vs_distance_heatmap_normalize(sd, matrix, output_path=None, n_bins_distance=50, global_bins=None, 
                                  n_bins_sttc=50, heatmap_vmax=None, title=None,
                                  normalization="density", subsample=None):
    """
    Generate a 2D heatmap over STTC and distance bins with various normalization options.
    
    Parameters:
        sd: Data object containing neuron data.
        matrix: STTC matrix.
        output_path: If provided, save the figure to this directory.
        n_bins_distance: Number of bins for the distance axis.
        global_bins: Custom bin edges for distance. If None, automatically computed.
        n_bins_sttc: Number of bins for the STTC axis.
        heatmap_vmax: Maximum value for the heatmap color scale (if None, determined automatically).
        title: Title of the plot.
        normalization: Method for normalization. Options are:
            "density" - converts counts to probability density,
            "minmax"  - scales counts to the [0,1] range,
            "zscore"  - standardizes counts (zero mean, unit variance),
            "raw"     - no normalization.
        subsample: If provided (as an integer), randomly select this many neuron pairs for the histogram.
            
    Returns:
        fig, ax: The matplotlib figure and axes objects.
    """
    sttc_matrix = matrix
    
    positions = get_neuron_positions(sd)
    if positions.size == 0:
        raise ValueError("No valid neuron positions found.")
    
    dist_matrix = compute_distance_matrix(positions)
    triu_indices = np.triu_indices_from(sttc_matrix, k=1)
    sttc_values = sttc_matrix[triu_indices]
    distance_values = dist_matrix[triu_indices]
    
    # If a subsample size is provided, randomly choose that many pairs.
    total_pairs = len(sttc_values)
    if subsample is not None:
        if subsample < total_pairs:
            sample_indices = np.random.choice(np.arange(total_pairs), size=subsample, replace=False)
            sttc_values = sttc_values[sample_indices]
            distance_values = distance_values[sample_indices]
    
    # Create bins for STTC and distance.
    sttc_bins = np.linspace(0, 1, n_bins_sttc + 1)
    if global_bins is None:
        global_bins = np.linspace(np.floor(np.nanmin(distance_values)),
                                  np.ceil(np.nanmax(distance_values)),
                                  n_bins_distance + 1)
    else:
        global_bins = np.array(global_bins)
    
    # Compute the 2D histogram.
    heatmap_matrix, _, _ = np.histogram2d(distance_values, sttc_values, bins=[global_bins, sttc_bins])
    
    # Apply normalization.
    if normalization == "density":
        # Calculate bin widths (assumed uniform).
        dx = global_bins[1] - global_bins[0]
        dy = sttc_bins[1] - sttc_bins[0]
        total_counts = heatmap_matrix.sum()
        if total_counts > 0:
            normalized_matrix = heatmap_matrix / (total_counts * dx * dy)
        else:
            normalized_matrix = heatmap_matrix
        cbar_label = 'Probability Density'
        
    elif normalization == "minmax":
        min_val = heatmap_matrix.min()
        max_val = heatmap_matrix.max()
        if max_val > min_val:
            normalized_matrix = (heatmap_matrix - min_val) / (max_val - min_val)
        else:
            normalized_matrix = heatmap_matrix
        cbar_label = 'Min-Max Normalized Count'
        
    elif normalization == "zscore":
        mean_val = heatmap_matrix.mean()
        std_val = heatmap_matrix.std()
        if std_val > 0:
            normalized_matrix = (heatmap_matrix - mean_val) / std_val
        else:
            normalized_matrix = heatmap_matrix
        cbar_label = 'Z-Score'
        
    elif normalization == "raw":
        normalized_matrix = heatmap_matrix
        cbar_label = 'Neuron Pair Count'
        
    else:
        raise ValueError(f"Unknown normalization method: {normalization}")
    
    # If no maximum value for color scale is provided, use the max of the normalized matrix.
    if heatmap_vmax is None:
        heatmap_vmax = normalized_matrix.max()
    
    # Plotting.
    fig, ax = plt.subplots(figsize=(8, 6))
    im = sns.heatmap(normalized_matrix, cmap="viridis", vmin=0, vmax=heatmap_vmax,
                     cbar_kws={'label': cbar_label})
    
    # Format the x-axis (STTC).
    num_xticks = 6
    xtick_positions = np.linspace(0.5, n_bins_sttc - 0.5, num_xticks)
    xtick_labels = ["{:.1f}".format(val) for val in np.linspace(0, 1, num_xticks)]
    ax.set_xticks(xtick_positions)
    ax.set_xticklabels(xtick_labels)
    
    # Format the y-axis (Distance).
    num_yticks = 5
    ytick_positions = np.linspace(0.5, n_bins_distance - 0.5, num_yticks)
    ytick_labels = ["{:.0f}".format(val) for val in np.linspace(global_bins[0], global_bins[-1], num_yticks)]
    ax.set_yticks(ytick_positions)
    ax.set_yticklabels(ytick_labels)
    
    ax.invert_yaxis()
    ax.set_xlabel("STTC Value")
    ax.set_ylabel("Distance (µm)")
    ax.set_title(f"STTC vs. Distance: {title}")
    
    if output_path:
        os.makedirs(output_path, exist_ok=True)
        fname = os.path.join(output_path, f"sttc_vs_distance_{title}.png")
        plt.savefig(fname)
        plt.close(fig)
        
    return fig, ax
```
